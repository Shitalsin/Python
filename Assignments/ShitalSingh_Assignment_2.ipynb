```python
import matplotlib.pyplot as plt

yearly_deliveries = df.groupby("Year")["Estimated_Deliveries"].sum()

plt.figure(figsize=(8,4))
plt.plot(yearly_deliveries.index, yearly_deliveries.values)
plt.title("Year-wise Tesla Deliveries")
plt.xlabel("Year")
plt.ylabel("Estimated Deliveries")
plt.show()
```

```python
yearly_price = df.groupby("Year")["Avg_Price_USD"].mean()

plt.figure(figsize=(8,4))
plt.plot(yearly_price.index, yearly_price.values)
plt.title("Average Tesla Price Trend")
plt.xlabel("Year")
plt.ylabel("Price (USD)")
plt.show()
```

```python
region_sales = df.groupby("Region")["Estimated_Deliveries"].sum()

region_sales.plot(kind="bar", figsize=(10,5))
plt.title("Region-wise Deliveries")
plt.show()
```

```python
model_sales = df.groupby("Model")["Estimated_Deliveries"].sum()

model_sales.plot(kind="bar", figsize=(10,5))
plt.title("Model-wise Deliveries")
plt.show()
```
